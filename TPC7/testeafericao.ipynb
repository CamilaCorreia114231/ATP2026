### tpc-1. Especifique as seguintes listas em compreensão:

#### a) Lista formada pelos elementos que não são comuns às duas listas:

````python
lista1 = [1, 2, 3, 4, 5]
lista2 = [4, 5, 6, 7, 8]  
def ncomuns (lista1, lista2):
    res = []
    for n in lista1:
        if not n in lista2:
            res.append(n)
    for n in lista2:
        if n not in lista1:
            res.append(n)
    return res

print(ncomuns(lista1, lista2))
````

#### b) Lista formada pelas palavras do texto compostas por mais de 3 letras:

````python
texto = """Vivia há já não poucos anos algures num concelho do Ribatejo 
    um pequeno lavrador e negociante de gado chamado Manuel Peres Vigário"""
lista = texto.split()
def mais3(lista, n):
    res = []
    for x in lista:
        if len(x) > n:
            res.append(x)
    return res

print(mais3(lista, 3))
````

#### c) Lista formada por pares do tipo (índice, valor) com os valores da lista dada:

````python
lista = ['anaconda', 'burro', 'cavalo', 'macaco']

def listaRes(lista):
    res = []
    i= 0
    for palavra in lista:
        res.append((i, palavra))
        i = i + 1
    return res

print (listaRes(lista))
````

#### a) Especifique uma função que dada uma string e uma substring não vazia, calcula  o número de vezes em que a substring aparece na string, sem que haja sobreposição de substrings:

````python
subs = input("Qual a substring que procuras na string principal?")

def strCount(s, subs):
    i = 0
    ct = 0 
    while i <= len(s) - len (subs):
        if s[i : i + len(subs)] == subs:
            ct = ct + 1
            i = i + len(subs)
        else:
            i = i + 1
    
    return ct

print(strCount("catcowcat", subs))

````

#### c) Especifique uma função que dado um número inteiro positivo, repetidamente adiciona os seus dígitos até obter apenas um dígito que é retornado como resultado:

````python
def reduxInt(n):
    while n >= 10:
        soma = 0
        for num in str(n):
            soma = soma + int(num)
        n = soma
    return n

print (reduxInt(25))
print (reduxInt(73))
print (reduxInt(676767))
````

#### d) Especifique uma função que recebe duas strings, `string1` e `string2`, e devolve o índice da primeira ocorrência de `string2` em `string1`, caso não ocorra nenhuma vez a função deverá retornar `-1`:

````python
def myIndexOf(s1, s2):
    ind = -1
    i = 0
    for i in range(len(s1)-len(s2)+1):
        if s1[i:i+len(s2)]== s2:
            ind = i
        i = i + 1
    return ind

print (myIndexOf("Hoje está um belo dia de sol!", "belo"))
print (myIndexOf("Hoje está um belo dia de sol!", "gato"))
print (myIndexOf("Hoje está um belo dia de sol!", "!"))
````

#### d) `insPost`, que acrescenta um novo post à rede social a partir dos parâmetros recebidos e devolve a nova rede social. 
    
O campo `id` devrá ser calculado a partir dos já existentes, por exemplo, se a rede tiver posts com id `p1`, `p2` e `p3`, o novo `id` deverá ser `p4`.

````python
def insPost(redeSocial, conteudo, autor, dataCriacao, comentarios):
    maiorid = 0
    for post in redeSocial:
        idatual = int(post["id"][1:])
        if idatual > maiorid:
            maiorid = idatual
    novoid = "p" + str(maiorid + 1)
    novopost = {
        "id": novoid,
        "conteudo": conteudo,
        "autor": autor,
        "dataCriacao": dataCriacao,
        "comentarios": []
    }
    redeSocial.append(novopost)
    return redeSocial
````

#### e)  `remPost`, que remove um post da rede, correspondente ao `id` recebido.

````python
def remPost(redeSocial, id):
    res = []
    for post in redeSocial:
        if id != post["id"]:
            res.append(post)
    return res
````

#### f) `postsPorAutor`, que devolve uma distribuição de posts por autor (à semelhança do que foi feito nas aulas).

````python
def postsPorAutor(redeSocial):
    distrib = {}
    for post in redeSocial:
        autor = post["autor"]
        if autor in distrib:
            distrib[autor] = distrib[autor] + 1
        else:
            distrib[autor] = 1
    return distrib
````

#### g) `comentadoPor`, que recebe um autor e devolve a lista de posts comentados por esse autor.

````python
def comentadoPor(redeSocial, autor):
    res = []
    for post in redeSocial:
        for comentario in post["comentarios"]:
            if comentario["autor"] == autor and post not in res:
                res.append(post)
    return res
````
